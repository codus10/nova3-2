## 🎯 PART 1. 핵심 통계 및 시각화 개념

2. 연속형 수치 데이터를 일정한 구간(Bin)으로 나누어 해당 구간에 속하는 데이터의 빈도를 막대 형태로 나타낸 그래프는 **( 히스토그 )** 입니다.
3. 위 2번 그래프의 계단식 분포를 수학적인 함수를 사용해 부드러운 연속 확률 밀도 곡선으로 스무딩(Smoothing)하는 기법을 알파벳 3글자로 **( KDE (Kernel Density Estimation) )**(이)라고 합니다.

**문제 6. [논리/심화] 코드로 이상치(Outlier) 직접 필터링하기**

* **목표:** 눈으로 본 이상치를 Pandas 데이터프레임 조작을 통해 실제로 추출해 냅니다.
* **조건:** `total_bill` 변수의 $Q1(25%)$과 $Q3(75\%)$ 값을 구함 -> $IQR$ 계산 -> $(Q3 + 1.5 \times IQR)$ 값을 초과하는 행(Row)만 필터링하여 출력.

* **✍️ 나의 AI 프롬프트:** (이 논리를 AI에게 어떻게 설명할까요?)

tips 데이터셋에서 total_bill 변수의 Q1(25%)와 Q3(75%)를 구하고, IQR을 계산한 뒤, Q3 + 1.5*IQR 값을 초과하는 행만 필터링해서 출력하는 파이썬 코드를 알려줘


* **💻 완성된 파이썬 코드:**
```python
import seaborn as sns
import pandas as pd

# 데이터셋 로드
tips = sns.load_dataset("tips")

# Q1, Q3 계산
Q1 = tips["total_bill"].quantile(0.25)
Q3 = tips["total_bill"].quantile(0.75)

# IQR 계산
IQR = Q3 - Q1

# 이상치 기준값
upper_bound = Q3 + 1.5 * IQR

# 이상치 필터링
outliers = tips[tips["total_bill"] > upper_bound]

print("이상치 행 개수:", len(outliers))
print(outliers)
```
---


````python
import seaborn as sns
import pandas as pd

# 데이터셋 로드
tips = sns.load_dataset("tips")

# Q1, Q3 계산
Q1 = tips["total_bill"].quantile(0.25)
Q3 = tips["total_bill"].quantile(0.75)

# IQR 계산
IQR = Q3 - Q1

# 이상치 기준값
upper_bound = Q3 + 1.5 * IQR

# 이상치 필터링
outliers = tips[tips["total_bill"] > upper_bound]

print("이상치 행 개수:", len(outliers))
print(outliers)
````

**문제 10. 피어슨 상관행렬 계산 (Correlation Matrix)**

* **목표:** 붓꽃(`iris`) 데이터의 수치형 변수들 간의 상관관계를 수치(상관계수)로 나타낸 표를 만드세요.
* **조건:** 문자열 변수(예: `species`)는 계산에서 제외할 것 / Pandas의 상관계수 계산 함수 사용.
* **💻 완성된 파이썬 코드:**
```python
import seaborn as sns
import pandas as pd

# 데이터셋 로드
iris = sns.load_dataset("iris")

# 수치형 변수만 선택 (species 제외)
numeric_iris = iris.drop(columns=["species"])

# 피어슨 상관행렬 계산
corr_matrix = numeric_iris.corr(method="pearson")

print("Iris Correlation Matrix (Pearson):")
print(corr_matrix)
```

````python
import seaborn as sns
import pandas as pd

# 데이터셋 로드
iris = sns.load_dataset("iris")

# 수치형 변수만 선택 (species 제외)
numeric_iris = iris.drop(columns=["species"])

# 피어슨 상관행렬 계산
corr_matrix = numeric_iris.corr(method="pearson")

print("Iris Correlation Matrix (Pearson):")
print(corr_matrix)
````

**문제 11. 히트맵(Heatmap) 시각화**

* **목표:** 문제 10에서 구한 숫자로 된 표를 직관적인 색상의 진하기로 시각화하세요.
* **조건:** 각 칸 안에 상관계수 소수점 숫자가 표시되도록 할 것 / 양의 상관관계와 음의 상관관계가 대비되는 색상 맵(예: `coolwarm`)을 적용할 것.
* **✍️ 나의 AI 프롬프트:** iris 데이터셋에서 species를 제외한 수치형 변수들의 피어슨 상관행렬을 heatmap으로 시각화하고, 각 칸에 상관계수 숫자가 표시되며 색상 맵은 coolwarm을 적용하는 파이썬 코드를 알려줘
* **💻 완성된 파이썬 코드:**
```python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
iris = sns.load_dataset("iris")

# 수치형 변수만 선택
numeric_iris = iris.drop(columns=["species"])

# 상관행렬 계산
corr_matrix = numeric_iris.corr(method="pearson")

# 히트맵 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Iris Correlation Matrix Heatmap")
plt.show()
```

````python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
iris = sns.load_dataset("iris")

# 수치형 변수만 선택
numeric_iris = iris.drop(columns=["species"])

# 상관행렬 계산
corr_matrix = numeric_iris.corr(method="pearson")

# 히트맵 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Iris Correlation Matrix Heatmap")
plt.show()
````

**문제 12. 전체 변수의 군집화 가능성 탐색 (Pairplot)**

* **목표:** 붓꽃 데이터를 품종별로 분류할 수 있는 핵심 변수가 무엇인지 찾기 위해, 모든 변수 간의 1:1 관계를 한 화면에 시각화하세요.
* **조건:** 붓꽃의 종(`species`)에 따라 점의 색상을 다르게 적용할 것.
* **💻 완성된 파이썬 코드:**
```python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
iris = sns.load_dataset("iris")

# Pairplot 시각화 (species별 색상 구분)
sns.pairplot(data=iris, hue="species")

plt.suptitle("Iris Pairplot by Species", y=1.02)
plt.show()
```

````python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
iris = sns.load_dataset("iris")

# Pairplot 시각화 (species별 색상 구분)
sns.pairplot(data=iris, hue="species")

plt.suptitle("Iris Pairplot by Species", y=1.02)
plt.show()
````
