## 🎯 PART 1. 핵심 전공 개념 확인

*두 데이터가 어떻게 같이 움직이는지 분석하는 필수 용어들을 점검해 봅니다.*

1. 한쪽 데이터가 커질 때 다른 쪽 데이터도 함께 커지거나 작아지는 등, '규칙적으로 같이 움직이는 경향'을 **(상관관계)**(Correlation)라고 합니다.
2. 기온이 올라갈 때 에어컨 판매량도 같이 올라가는 것처럼, 두 데이터가 같은 방향으로 움직이는 것을 **( 양 )** 의 상관관계라고 합니다. (산점도에서 우상향)
3. 기온이 올라갈 때 난로 판매량은 떨어지는 것처럼, 두 데이터가 반대 방향으로 움직이는 것을 **(음)** 의 상관관계라고 합니다. (산점도에서 우하향)
4. 두 데이터가 얼마나 강하게 엮여 있는지를 보여주는 숫자인 '상관계수'는 항상 **( -1 )** 부터 **( 1 )** 사이의 값을 가집니다.
5. 상관계수가 **( 0 )** 에 가까울수록 두 데이터는 아무런 관계가 없는 완벽한 남남(무상관)이라는 뜻입니다.
6. [가장 중요 ⭐] A와 B가 같이 움직인다고 해서 "A 때문에 B가 일어났다"고 단정 지을 수 없습니다. 즉, "상관관계는 **( 인과관계 )** (Causation)를 의미하지 않는다"는 분석가의 제1원칙입니다.
7. 아이스크림 판매량이 늘어날 때 상어 피해가 증가하는 현상의 진짜 원인은 '여름(폭염)'이었습니다. 이렇게 둘을 동시에 움직이게 만든 숨겨진 진짜 원인을 **( 제3의변수 )** 의 변수라고 부릅니다.
8. 판다스 데이터프레임에서 모든 숫자형 변수들끼리의 상관계수를 한 번에 짝지어 계산해 주는 메서드는 `df. corr()` 입니다.
9. 복잡한 상관계수 표를 볼 때, 숫자의 크기에 따라 색깔의 진하기를 다르게 칠해서 한눈에 관계의 강도를 파악하게 해주는 그래프를 **( 히트맵 )** (Heatmap)이라고 합니다.
10. 두 데이터의 관계를 파악하기 위해 가로축과 세로축에 점을 찍어 그리는 그래프는 **( 산점도 )** (Scatter plot)입니다.

* **나의 프롬프트:** > "파이썬 판다스로 '공부시간'과 '시험점수' 열을 가진 가상 데이터프레임 `df`를 10줄 만들어줘. 공부시간이 길수록 시험점수가 높아지는 '양의 상관관계'를 가지도록 수치를 적당히 맞춰줘."

**미션 2. 상관행렬(Correlation Matrix) 한 번에 계산하기**

* **상황:** 데이터프레임 `df`에 들어있는 모든 숫자 데이터들끼리의 상관계수 표를 화면에 출력하고 싶습니다.
* **나의 프롬프트:** > "파이썬 판다스로 데이터프레임 df 안의 모든 숫자형 열들끼리 상관계수를 계산해서 상관행렬(Correlation Matrix) 표를 출력해줘. df.corr() 메서드를 사용해서 결과를 보여줘."


````python
import pandas as pd

# 예시 데이터 (앞에서 만든 df 사용 가능)
data = {
    "공부시간(시간)": [1, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5, 6],
    "시험점수(점)": [55, 60, 65, 70, 74, 78, 82, 88, 92, 96]
}
df = pd.DataFrame(data)

# 모든 숫자형 열들의 상관계수 행렬 출력
print(df.corr())
````

**미션 3. 산점도(Scatter plot)로 눈으로 확인하기**

* **상황:** `seaborn` 라이브러리를 사용해서, x축을 '공부시간', y축을 '시험점수'로 설정하여 점을 찍어보는 그래프를 그리고 싶습니다.
* **나의 프롬프트:** > "파이썬에서 seaborn을 사용해 데이터프레임 df의 산점도를 그려줘. x축은 '공부시간', y축은 '시험점수'로 설정해서 점들을 찍어 보여줘"


````python
import seaborn as sns
import pandas as pd

# 예시 데이터
scores = {
    "공부시간": [1, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5, 6],
    "시험점수": [55, 60, 65, 70, 74, 78, 82, 88, 92, 96]
}
df = pd.DataFrame(scores)

# 산점도 그리기
sns.scatterplot(data=df, x='공부시간', y='시험점수')
````

**미션 4. 색깔로 직관적인 히트맵(Heatmap) 그리기**

* **상황:** 미션 2에서 계산한 상관계수 표를 `seaborn`의 히트맵으로 시각화하고 싶습니다. 칸마다 상관계수 숫자가 보이게 하고, 테마는 파란색과 빨간색이 대비되는 `coolwarm`으로 하고 싶습니다.
* **나의 프롬프트:** > "파이썬에서 seaborn을 사용해 데이터프레임 df.corr() 결과를 히트맵으로 시각화해줘. 칸마다 상관계수 숫자가 표시되도록 annot=True 옵션을 켜고, 색상 테마는 coolwarm을 적용해줘"


````python
import seaborn as sns
import matplotlib.pyplot as plt

# 상관계수 행렬 계산
corr_matrix = df.corr()

# 히트맵 시각화
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()
````

**미션 5. 추세선이 포함된 산점도(Regplot) 그리기**

* **나의 프롬프트:** > "파이썬에서 seaborn의 regplot을 사용해 데이터프레임 df의 산점도를 그려줘. x축은 '공부시간', y축은 '시험점수'로 설정하고, 점들 사이를 관통하는 선형 회귀선(추세선)도 함께 표시해줘"

---

````python
import seaborn as sns
import pandas as pd

# 예시 데이터
scores = {
    "공부시간": [1, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5, 6],
    "시험점수": [55, 60, 65, 70, 74, 78, 82, 88, 92, 96]
}
df = pd.DataFrame(scores)

# 산점도 + 추세선 그리기
sns.regplot(data=df, x='공부시간', y='시험점수')
````
