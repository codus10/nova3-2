* **상황:** 데이터프레임 `data`에서 원인이 되는 '광고비' 열은 2차원 배열 형태인 독립변수 `X`로, 결과가 되는 '매출액' 열은 종속변수 `Y`로 나누어 저장하고 싶습니다.

````python
from sklearn.linear_model import LinearRegression
import pandas as pd
from sklearn.datasets import load_diabetes

# 데이터셋 불러오기
diabetes = load_diabetes()
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Feature와 Target 설정
X = df[['bmi']]   # 독립변수: 체질량지수
Y = df['s5']      # 종속변수: 특정 혈청 지표

# 모델 초기화
ai_model = LinearRegression()

# 모델 학습 (Fit)
ai_model.fit(X, Y)

# 학습 결과 확인
print("기울기(Weight):", ai_model.coef_[0])
print("절편(Bias):", ai_model.intercept_)
````

**미션 2. 수학 공식(회귀식) 도출하기**

* 학습된 모델에서 기울기(`coef_`)와 절편(`intercept_`)을 추출하세요.
* 이 AI 모델이 찾아낸 당뇨병 예측 공식은 어떻게 되나요? 소수점 첫째 자리까지만 반올림하여 적어보세요.
👉 **s5 = (0.9) $\times$ bmi + (0.0)**

````python
weight = ai_model.coef_[0]      # 기울기
bias = ai_model.intercept_      # 절편

print("기울기:", weight)
print("절편:", bias)
````
